# Introduction

- **Project setup**:  
  - Four different biological datasets.  
  - Each dataset contains two different types of cells.  
  - Each uses two different RNA sequencing methods.  

- **Dataset structure**:  
  - Features = genes (rows).  
  - Samples = cells (columns).  
  - Each cell has a label: **hypoxic** or **normoxic** environment.  

- **Goal**:  
  - Build a model for each dataset.  
  - The model should predict the correct label (hypoxic/normoxic) based on gene expression.  

- **Case studies**:  
  - **HCC1806**: epithelial cell line, mammary gland, 60-year-old Black female patient, acantholytic squamous cell carcinoma (ASCC).  
  - **MCF-7**: breast cancer cell line, 69-year-old female patient.  

- **Sequencing methods**:  
  - **SmartSeq**: deeper sequencing, richer per-cell information, but not suitable for large sample sizes.  
  - **DropSeq**: more shallow sequencing, less per-cell information, but suitable for larger datasets.  

- **Summary of sequencing trade-offs**:  
  - DropSeq → more samples, less detail per cell.  
  - SmartSeq → fewer samples, more complete RNA sequence per sample.  

- **Planned approach**:  
  1. Data exploration.  
  2. Preprocessing to prepare data for training.  
  3. Unsupervised learning.  
  4. Supervised learning.  
  5. Model selection (best model per dataset).  
  6. Final predictions on the test set.  
  7. Conclusion of results.
